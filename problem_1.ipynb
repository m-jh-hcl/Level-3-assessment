```
"""
Problem Statement 1

A company wants to predict employee productivity scores to improve workforce planning and training programs.
You are hired as a Data Scientist to build a multivariate linear regression model that predicts an employee's
Productivity Score based on multiple work-related factors.

Experience (yrs),Training Hours,Working Hours,Projects,Productivity Score
2,40,38,3,62
5,60,42,6,78
1,20,35,2,55
8,80,45,8,88
4,50,40,5,72
10,90,48,9,92
3,30,37,4,65
6,70,44,7,82
7,75,46,7,85
2,25,36,3,60

Interpretation

Which factor most strongly impacts productivity?
How does training affect productivity?
Should the company increase training hours or working hours?
What happens if Working Hours increase beyond optimal limits?
Can productivity ever decrease with more experience?
How would you detect overfitting in this model?
Suggest one new feature to improve prediction accuracy.
"""
```

```
import pandas as pd

dataset_file = 'ps-1.csv'

# Prepare the dataset
try:
    df = pd.read_csv(dataset_file)
    print("Dataset loaded successfully.")
except FileNotFoundError:
    print("File not found!")

df.head()
```

```
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler

# Split features (X) and target (y)
X = df[['Experience (yrs)', 'Training Hours', 'Working Hours', 'Projects']]
y = df['Productivity Score']

## Not splitting the data into train-test sets as the number of samples are small

# Scale the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Build and train the Linear Regression Model
model = LinearRegression()
model.fit(X_scaled, y)

# Extract coefficients and intercept
coefficients = pd.Series(model.coef_, index=X.columns)
intercept = model.intercept_

# Results
print("--- Model Results ---")
print(f"Intercept: {intercept:.2f}")
print("Coefficients:")
print(coefficients)
```

```
"""
Which factor most strongly impacts productivity?

Projects has the largest coefficient by far (10.55). This has the most significant gain compared to any other factor.
"""
```

```
"""
How does training affect productivity?

Training has a positive impact (coefficient 0.89). Its individual impact is smaller than that of projects or working hours.
"""
```

```
"""
Should the company increase training hours or working hours?

Based on these coefficients, the company should increase working hours. The coefficient for Working Hours (3.54)
is much higher than for Training Hours (0.89). Therefore according to this specific model, Working Hours
contribute nearly four times more to the productivity score than Training.
"""
```

```
"""
What happens if Working Hours increase beyond optimal limits?

In this linear model, the score will continue to increase indefinitely. Because the coefficient is a fixed positive number (3.54),
the model doesn't consider human fatigue. It does not account for the point where more hours lead to a drop in quality.

"""
```

```
"""
Can productivity ever decrease with more experience?

Yes, for this model. The coefficient for Experience is -2.68. This implies that Experience has a negative impact on productivity.
"""
```

```
"""
How would you detect overfitting in this model?

For only 10 samples of data, if the R-Squared value is very high (0.997 here), the model has likely memorized the specific details
of these 10 employees rather than learning a general rule that would apply to new samples.
"""
```

```
r_squared = model.score(X_scaled, y)
r_squared
```
